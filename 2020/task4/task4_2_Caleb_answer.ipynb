```python
#Task 4.2
class Person:
  def __init__(self, full_name, date_of_birth):
    self.full_name = full_name
    self.date_of_birth = date_of_birth

  def is_adult(self):
    birth_year = int(self.date_of_birth[:4])
    today_year = 2026
    age = today_year - birth_year

    if age > 18:
      return True
    else:
      return False

  def screen_name(self):
    name = self.full_name.replace(" ", "")
    month = self.date_of_birth[5:7]
    day = self.date_of_birth[8:10]

    return name + month + day
###############################################
class Staff(Person):
  def is_adult(self):
    return True

  def screen_name(self):
    return super().screen_name() + "Staff"

class Student(Person):

  def is_adult(self):
    return False

###############################################
#Main Program
p1 = Person("John Tan", "2000-06-01")

print(p1.is_adult())
print(p1.screen_name())
```

```python
#Task 4.2 part 2
import sqlite3

conn = sqlite3.connect("school.db")
cursor = conn.connect

with open("people.txt", "r") as f:
  for line in f:
    line = line.strip()
    details = line.split(",")

    if details[2] == "Staff":
      person = Staff(details[0], details[1])
    elif details[2] == "Student":
      person = Student(details[0], details[1])
    else:
      person = Person(details[0], details[1])

  #file automatically closes

    screen = person.screen_name()

    if person.is_adult():
      adult = 1
    else:
      adult = 0

    cursor.execute("""
      INSERT INTO People
      (FullName, DateofBirth, ScreenName, IsAdult)
      Values (?, ?, ?, ?)""",
      (person.full_name, person.date_of_birth, screen, adult)
      )

  conn.commit()
  conn.close()


```
