```python
# Task 1.1
import pymongo, json
client = pymongo.MongoClient("127.0.0.1", 27017)
db = client["ParcelDB"]
parcels = db["Parcels"]
```

```python
# Task 1.2
def loadparcels():
  with open("TASK_2.JSON", "r") as file:
    parcel_data = json.load(file)
  #file automatically closed

  #Clear existing documents so repeated testing does not create duplicates.
  parcels.delete_many({})

  if len(parcel_data) > 0:
    parcels.insert_many(parcel_data)
    print("Parcel records loaded sucessfully.")
  else:
    print("No parcel records found in JSON file.")
```

```python
# Task 1.3
def displayParcel(parcelID):
  parcel = parcels.find_one({"parcelID": parcel ID})

  if parcel is not None:
    print("---Parcel Found---")
    print("Parcel ID :", parcel["parcelID"])
    print("Recipient :", parcel["recipient"])
    print("Priority :", parcel["priority"])
    print("Status :", parcel["status"])
  else:
    print("---Parcel Not Found---")
```

```python
# Task 1.4
def priorityParcels(priorityType):
  results = parcels.find({"priority": priorityType})
  count = 0
  print(f"{'Parcel ID':<12}{'Recipient':<20}")
  for parcel in results:
    print(f"{parcel['parcelID':<12]}{parcel['reciptient']<:20}")
    count += 1
  print("Total no. of parcels:", count)
```

```python
# Main program for testing
if__name__ == "__main__":
  loadparcels()
  print()

  displayParcel("P1001")
  print()

  priorityParcels("Standard") #Task 1.5
  print()
```
