The tree is implemented using Object-Oriented Programming (OOP).

The class `Tree` contains three properties:

- `left_pointer` points to the left subtree
- `right_pointer` points to the right subtree
- `data` is the data in the node.

The class `Tree` contains the following methods:

- a constructor to set the left pointer and right pointer to None, and the data to its parameter
- a recursive method to take the parameter and store it in the correct position in the tree
- a recursive method to use in-order traversal to output the data in the tree

````python
#Task 3.1
class Tree:
    #constructor
    def __init__(self, data):
        self.left_pointer = None
        self.right_pointer = None
        self.data = data
        
    def getLeft(self):
        return self.left_pointer
        
    def setLeft(self, left):
        self.left_pointer = left
        
    def getRight(self):
        return self.right_pointer
        
    def setRight(self, right):
        self.right_pointer = right
        
    def getdata(self):
        return self.data
        
    def setdata(self, data):
        self.data = data
        
    #methods
    def insert(self, data):
        if data < self.data:
            if self.left_pointer == None:
                self.left_pointer = Tree(data)
            else:
                self.left_pointer.insert(data)
        
        elif data > self.data:
            if self.right_pointer == None:
                self.right_pointer = Tree(data)
            else:
                self.right_pointer.insert(data)
                
    def inorder(self):
        if self.left_pointer != None:
            self.left_pointer.inorder()
            
        print(self.data)
        
        if self.right_pointer != None:
            self.right_pointer.inorder()
            
            
    def postorder(self):
        if self.left_pointer != None:
            self.left_pointer.postorder()
        
        if self.right_pointer != None:
            self.right_pointer.postorder()
            
        print(self.data)
            
        
                
                
#Main program
import random

BST = Tree(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))
BST.insert(random.randint(0, 999))

print("In-order traversal")
BST.inorder()
print("-----------------------")
print("Post-order traversal")
BST.postorder()
````
