**Task 3.1**

Write the `LinkedList` class in Python. Use of a simple Python list is not sufficient. Include the following methods:

- `insert(integer_value)` inserts the `integer_value` at the beginning (head) of the list
- `delete(integer_value)` attempts to delete `integer_value` from the list; if the item was not present, return `None`
- `search(integer_value)` returns a Boolean value: `True` if `integer_value` is in the list, `False` if not in the list
- `count()` should return the number of elements in the list, or zero if empty
- `to_String()` should return a string containing a suitably formatted list with the elements separated by a comma and a space, with square brackets at either end, e.g. in the form:

  `[11, 2, 7, 4]`

  **[8]**

Test `LinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list. **[3]**


````python
# Task 3.1
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
class LinkedList:
    def __init__(self):
        self.head = None
        
    def gethead(self):
        return self.head
    
    def sethead(self, head):
        self.head = head
        
    def isEmpty(self):
        if self.head == None:
            return True
        else:
            return False
        
    def insert(self, data): #insert at head
        newNode = Node(data)
        
        if self.isEmpty():
            self.head = newNode
        else:
            newNode.next = self.head
            self.head = newNode
            
    def delete(self, target):
        if self.isEmpty():
            return None
        
        if self.head.data == target:
            self.head = self.head.next
            return
            
        current = self.head
        while current.next != None and current.next.data != target:
            current = current.next
            
        if current.next == None:
            return None
        
        current.next = current.next.next
        
    def search(self, target):
        if self.isEmpty():
            return False
        
        if self.head.data == target:
            return True
        
        current = self.head
        while current.next != None and current.next.data != target:
            current = current.next
            
        if current.next == None:
            return False
        else:
            return True
        
    def count(self):
        count = 0 
        if self.isEmpty():
            return count
        
        current = self.head
        while current != None:
            count += 1 
            current = current.next
            
        return count
            
    def to_String(self):
        output = []
        
        if self.isEmpty():
            return "Its empty..."
        
        current = self.head
        while current != None:
            value = str(current.data)
            value = value.strip("\n")
            value = int(value)
            output.append(value)
            current = current.next
            
        return output
    
    
#Main Program
LL = LinkedList()
with open("Task3data.txt", "r") as file:
    for i in file:
        LL.insert(i)
    #automatically close file
    
print(LL.to_String())
````

Write a Python subclass `SortedLinkedList` using `LinkedList` as its superclass.

The insert method in the `SortedLinkedList` subclass should ensure that the elements are stored in ascending order. **[5]**

Test `SortedLinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list.

Print the result of searching the `SortedLinkedList` for the value 94. **[2]**


````python
# Task 3.2
class SortedLinkedList(LinkedList):
    
    def insert(self, value):
        newNode = Node(value)
        
        if self.isEmpty(): #empty list
            self.head = newNode
            return
        
        if value < self.head.data:
            newNode.next = self.head
            self.head = newNode
            return
        
        current = self.head
        while current.next != None and current.next.data < value:
            current = current.next
        
        newNode.next = current.next
        current.next = newNode


#Main program
SLL = SortedLinkedList()
with open("Task3data.txt", "r") as file:
    for i in file:
        SLL.insert(i)
    #automatically close file
    
print(SLL.to_String())
````

Write a Python subclass `Queue` using `LinkedList` as its superclass.

Additional `enqueue` and `dequeue` methods are to be defined on the `Queue` class:

- `enqueue(integer_value)` will insert `integer_value` to the end of the queue
- `dequeue()` will return the first element in the queue. If the queue is empty, return `None`. **[6]**

Test `Queue` by using the data in the file `Task3data.txt`. Print the first five elements to be dequeued from the list. **[1]**


````python
# Task 3.3
class Queue(LinkedList):
    def enqueue(self, value):
        newNode = Node(value)
        
        if self.isEmpty():
            self.head = newNode
            return
        
        current = self.head
        while current.next != None:
            current = current.next
            
        current.next = newNode
        return
    
    def dequeue(self):
        if self.isEmpty():
            return None
        else:
            value = self.head.data
            self.head = self.head.next
            return print(value)
    
    
#Main program
Que = Queue()
with open("Task3data.txt", "r") as file:
    for i in file:
        Que.enqueue(i)
    #automatically close file
    
print(Que.to_String())
print()
Que.dequeue()
Que.dequeue()
Que.dequeue()
Que.dequeue()
Que.dequeue()

print(Que.to_String())
````
